```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    RobustScaler,
    StandardScaler,
    OneHotEncoder,
    OrdinalEncoder
    )
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    confusion_matrix,recall_score,precision_score,f1_score,accuracy_score
)
```

```python
df = pd.read_csv('loan_data.csv')
df.head()
```

```python
X = df.drop(columns='loan_status')
y = df['loan_status']
```

```python
num_cols = X.select_dtypes(include='number').columns
cat_cols = X.select_dtypes(include='object').columns
```

```python
xtrain,xtest,ytrain,ytest = train_test_split(
    X,y,train_size=0.8,random_state=42
)
```

```python
X[cat_cols].nunique()
```

```python
X['person_education'].unique()
```

```python
r_columns = ['person_age', 'person_income', 'person_emp_exp', 'loan_amnt',
    'loan_percent_income','credit_score']
s_columns = ['cb_person_cred_hist_length','loan_int_rate']
ord_cols = ['person_education']
onehot_cols = cat_cols.drop('person_education')
```

```python
ord_encoder = OrdinalEncoder(
    categories=[['High School', 'Bachelor','Master', 'Associate', 'Doctorate']],
    handle_unknown='use_encoded_value',
    unknown_value=-1
)
one_hot_encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown='ignore'
)
robust_scaler = RobustScaler()
standard_scaler = StandardScaler()
```

```python
#scaling techniques
#robust_scaler
xtrain[r_columns] = robust_scaler.fit_transform(xtrain[r_columns])
xtest[r_columns] = robust_scaler.transform(xtest[r_columns])
#standard_scaler
xtrain[s_columns] = standard_scaler.fit_transform(xtrain[s_columns])
xtest[s_columns] = standard_scaler.transform(xtest[s_columns])
xtrain
```

```python
#encoding_technique
#ordinal_encoder
xtrain[ord_cols] = ord_encoder.fit_transform(xtrain[ord_cols])
xtest[ord_cols] = ord_encoder.transform(xtest[ord_cols])
```

```python
xtrain
```

```python
#onehot_encoder

onehot_cols
```

```python
onehot_cols
```

```python
xtrain_encoded_values = one_hot_encoder.fit_transform(xtrain[onehot_cols])
encoded_cols = one_hot_encoder.get_feature_names_out()
xtrain[encoded_cols] = xtrain_encoded_values
xtrain.drop(onehot_cols,axis=1,inplace=True)
```

```python
xtest_encoded_values = one_hot_encoder.fit_transform(xtest[onehot_cols])
xtest[encoded_cols] = xtest_encoded_values
xtest.drop(onehot_cols,axis=1,inplace=True)
```

```python
xtrain
```

```python
xtest
```

```python
model = LogisticRegression()
model.fit(xtrain,ytrain)
```

```python
y.value_counts()
```

```python
ytrain.value_counts()
```

```python
ytrain_pred = model.predict(xtrain)
ytest_pred = model.predict(xtest)
c_m = confusion_matrix(ytrain,ytrain_pred)
c_m
```
